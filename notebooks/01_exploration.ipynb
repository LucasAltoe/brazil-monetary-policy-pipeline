```python
import pandas as pd
from pathlib import Path
```

```python
RAW = Path('../data/raw')
```

```python
selic = pd.read_csv(RAW / 'selic.csv', sep=';', decimal=',')
selic.head()
```

```python
selic.dtypes
```

```python
ipca = pd.read_csv(RAW / 'ipca.csv', sep=';', decimal=',')
ipca.head()
```

```python
ipca.dtypes
```

```python
selic = selic.rename(columns={
    "valor": "selic"
})
```

```python
selic.head()
```

```python
ipca = ipca.rename(columns={
    "valor":"ipca"
})
```

```python
ipca.head()
```

```python
selic['data'] = pd.to_datetime(selic['data'], format='%d/%m/%Y')
```

```python
selic.head()
```

```python
ipca['data'] = pd.to_datetime(ipca['data'], format='%d/%m/%Y')
```

```python
ipca.head()
```

```python
selic_ipca_comparative = selic.merge(
    right=ipca,
    on= 'data',
    how='inner')
```

```python
print(selic_ipca_comparative.shape)
print(selic_ipca_comparative.isna().sum())
```
